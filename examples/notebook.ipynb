# seekrit in a notebook

One call loads this token's environment into `os.environ`. Run the first
code cell: if `SEEKRIT_TOKEN` isn't set in the kernel's environment you'll
get a password prompt, which keeps the token out of this file.


```python
import seekrit

seekrit.load()
```

The output above lists **names**, never values — that's what gets saved
into the `.ipynb`. The values are in the environment, where your libraries
already look for them.


```python
import os

# a name, not a value: printing a secret would write it into this notebook
print("DATABASE_URL set:", "DATABASE_URL" in os.environ)
```

Re-running the `seekrit.load()` cell refreshes every name, so a rotated
secret takes effect without restarting the kernel.

Full guide: https://seekrit.dev/docs/guides/notebooks
